```python
import networkx as nx
```

```python
#Empty graph
G = nx.Graph()
```

```python
#Add nodes
G.add_node(1)
G.add_nodes_from([2,3])
G.add_nodes_from([(4,{"color": "red"}), (5,{"color": "green"})])
```

```python
#Incorporate nodes from another graph
H = nx.path_graph(10)
G.add_nodes_from(H)
```

```python
#Add edges
G.add_edge(1, 2)
e = (2, 3)
G.add_edge(*e)
G.add_edges_from([(1,2), (2,3)])
```

```python
#Incorporate edges from another graph
G.add_edges_from(H.edges)
```

```python
#Remove all nodes and edges
G.clear()
```

```python
#Add nodes and edges
G.add_edges_from([(1, 2), (1, 3)])
G.add_node(1)
G.add_edge(1, 2)
G.add_node("spam")        # adds node "spam"
G.add_nodes_from("spam")  # adds 4 nodes: 's', 'p', 'a', 'm'
G.add_edge(3, 'm')
```

```python
#Get graph information
G.number_of_nodes(), G.number_of_edges()
```

```python
DG = nx.DiGraph()
DG.add_edge(2, 1)   # adds the nodes in order 2, 1
DG.add_edge(1, 3)
DG.add_edge(2, 4)
DG.add_edge(1, 2)
assert list(DG.successors(2)) == [1, 4]
assert list(DG.edges) == [(2, 1), (2, 4), (1, 3), (1, 2)]
```

```python
#Visualization with pyvis
from pyvis.network import Network

net = Network(notebook=True)
net.from_nx(G)
net.toggle_physics(True)
net.show('net.html')
```

```python
#Directed graph
DG = nx.DiGraph()
DG.add_weighted_edges_from([('1', '2', 0.5), ('3', '1', 0.75),('3','3',1)])
```

```python
dnet = Network(notebook=True,directed=True)
dnet.from_nx(DG)
dnet.toggle_physics(True)
dnet.show('dnet.html')
```
